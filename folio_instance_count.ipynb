## 2. Get the count of Instance records

`GET /instance-storage/instances?limit=0` returns no records but an exact `totalRecords` value — this is documented FOLIO/RMB behavior for `limit=0`.

```python
if not session:
    print("No session available — check the login output above before running this cell.")
else:
    instances_url = f"{OKAPI_URL}/instance-storage/instances"

    # session already carries the X-Okapi-Tenant and X-Okapi-Token headers
    # (set inside folio_login, in folio_auth.ipynb), plus the auth cookie,
    # so no headers need to be set again here.
    params = {
        "limit": 0,
        "totalRecords": "exact",
    }

    count_response = session.get(instances_url, params=params)

    print(f"Status code: {count_response.status_code}")

    if count_response.status_code == 200:
        data = count_response.json()
        total = data.get("totalRecords")
        print(f"Total Instance records: {total}")
    else:
        print("Request failed.")
        print("Response body:", count_response.text)
```
